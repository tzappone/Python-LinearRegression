```python
import pandas as pd
import csv

car_data = pd.read_csv(r'car_data.csv')
```

```python
# Question 1 - using car_engine_displacement to predict car_price

# Linear regression assessment statistics between car_engine_displacement and car_price
from scipy import stats
question1 = stats.linregress(x = car_data.car_engine_displacement, y= car_data.car_price)
# Printing both the full linear regression model as well as the outlined assessment statistics from the model
print(question1)
print(f'R-Squared: {(question1.rvalue ** 2):.4f}')
print(f'P-value: {question1.pvalue:.4f}')
print(f'Slope: {question1.slope:.4f}')
print(f'Intercept: {question1.intercept:.4f}')

#Question 1 - Graph
import matplotlib
import seaborn as sns
sns.regression.regplot(x = car_data.car_engine_displacement, y= car_data.car_price)

#Question 1 - Predicting the price of a car with 2,500 cubic centimeters of displacement
car_engine_dis = 2500
predicted_price1 = (car_engine_dis * question1.slope) + question1.intercept
print(f'The predicted price of a car with 2,500 cubic centimeters of displacement is: {predicted_price1:.2f}')


```
